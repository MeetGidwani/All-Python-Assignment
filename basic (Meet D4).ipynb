```python
Name = str(input("Enter your Name:"))
Roll_no = int(input("Enter roll no"))
phy = int(input("Enter your Physics Marks"))
chem = int(input("Enter your chemistry Marks"))
mat = int(input("Enter your maths Marks"))

percentage =(mat + phy + chem)/3 *100
print("\nStudent name:" , Name)
print("Roll number:" , Roll_no)
print("Percentage:", percentage)
#Find highest marks
if mat>= phy and mat >=chem:
    highest ="Maths"
elif phy >= mat and phy>= chem:
    highest =" Physics"
else:
    highest ="Chemistry"

#for lower marks
if mat<= phy and mat <=chem:
    lowest ="Maths"
elif phy <= mat and phy<= chem:
    lowest =" Physics"
else:
    lowest ="Chemistry"
print("Subject with highest marks:" , highest)
print("Subject with lower marks:" , lowest)
```
